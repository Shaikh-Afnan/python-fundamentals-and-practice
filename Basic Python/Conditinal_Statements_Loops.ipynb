# Conditional Statements and Loops in Python
## What are Conditional Statements?
Definition:

Conditional statements allow a program to make decisions and execute different blocks of code based on whether a condition is True or False.

They control the flow of execution in a program.

Real-Life Example:

If it is raining → take an umbrella
Else → wear sunglasses

## Types of Conditional Statements in Python

## (a) if Statement
Definition:

Executes a block of code only if the specified condition is True.

Syntax:

if condition: 
>     # code to execute

```python
age = 20

if age >= 18:
    print("You are eligible to vote.")
```

## (b) if-else Statement
Definition:

Executes one block of code if the condition is True, otherwise executes another block.

Syntax:

if condition:  
>    \# True block

else:   
>    \# False block

```python
num = 7

if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

## (c) if-elif-else Ladder
Definition:

Used when multiple conditions need to be checked. Python evaluates conditions from top to bottom and executes the first True block.

Syntax:  

if condition1:  
>    pass
  
elif condition2:  
>    pass

else:   
>    pass  

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

## (d) Nested if Statement
Definition:

An if statement inside another if statement is called a nested if.

```python
age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to vote")
```

## (e) Short-Hand if (Ternary Operator)
Definition:

A one-line conditional statement used to simplify simple if-else conditions.

Syntax:

value_if_true if condition else value_if_false

```python
num = 10
result = "Even" if num % 2 == 0 else "Odd"
print(result)
```

## Important Interview Points

- Python uses indentation instead of {}
- Conditions must return True or False
- You can combine conditions using:

`and`

`or`

`not`

Example:


```python
age = 22
salary = 30000

if age > 18 and salary > 25000:
    print("Eligible for loan")
```

## What is a Loop?
Definition:

A loop is a control structure that repeatedly executes a block of code until a specified condition is met.

### Why Loops Are Needed:

- Avoid repetition

- Reduce code length

- Automate repetitive tasks

### Types of Loops in Python

Python supports two main loops:

`1` while loop

`2` for loop 

Note :- Python does not have a built-in do-while loop.

## 1. while Loop

Definition:

The while loop executes a block of code as long as the given condition is True.

Syntax:

while condition:
>    \# loop body


Example :

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
# reverse while loop : 

i = 5

while i >= 1:
    print(i)
    i -= 1
```

### (a) Infinite while Loop
Definition:

A loop that never ends because the condition always remains True.

syntax : 

while True:
>  \# code

Example : 

while True:
>    print("Running...")

Note : Avoid unless required.

### (b) else with Loop

The else block in a loop executes only when the loop finishes normally without being interrupted by a break statement.

Example : 

```python
#  while Loop with else
count = 1

while count <= 3:
    print(count)
    count += 1
else:
    print("While loop completed")
```

```python
# for Loop with else

for i in range(3):
    print(i)
else:
    print("Loop finished successfully")
```

## 2. for Loop
Definition:

The for loop iterates over a sequence (list, tuple, string, dictionary, set, range) and executes code for each element.

Syntax:

for variable in iterable:

>    \# loop body

```python
for i in range(5):
    print(i)
```

```python
# reverse for loop

for i in range(8, 0 , -1):
    print(i)
```

### Common Iterables Used with for
List:

```python
for item in [10, 20, 30]:
    print(item)
```

String : 

```python
string = 'Python'

for ch in string:
    print(ch)
```

Dictionary : 

```python
student = {"name":"Afnan", "age":22}

for key, value in student.items():
    print(key, value)
```

## Loop Control Statements

These change the normal flow of loops.

### (a) break

Definition:

Terminates the loop immediately.

```python
for i in range(10):
    if i == 5:
        break     # breaks the loop when the value of i equals 5
    print(i)
```

```python
i = 0

while i <= 10:
    if i == 5:
        break     # breaks the loop when the value of i equals 5
    print(i)
    i += 1
```

### (b) continue
Definition:

Skips the current iteration and moves to the next one.

```python
for i in range(5):
    if i == 2:
        continue       # when the value of i equals 2 the iteration is skipped
    print(i)
```

```python
i = 1

while i <= 5:
    if i == 2:      # 2 is skipped
        i += 1
        continue     # when the value of i equals 2 the iteration is skipped
    print(i)
    i += 1
```

### (d) pass
Definition:

Acts as a placeholder when a loop body is syntactically required but no action is needed.

```python
for i in range(3):
    pass

# nothing happens until the pass statement is removed
# it can be used inside loops, conditional statements, functions, classes, methods
```

### Using pass statement in while loop

count = 0

while count < 5:
>    pass


Important:
This creates an infinite loop because count is never updated.




```python
# Practical Example (Avoid Infinite Loop):
count = 0

while count < 5:
    count += 1
    pass

print("Loop finished")
```

### (e) Nested Loops
Definition:

A loop inside another loop is called a nested loop.

```python
# Using nested for loop

for i in range(3):
    for j in range(2):
        print(i, j)
```

Use Case:

- Pattern printing

- Matrix operations

- Comparing elements

##  Looping with Index

###  1. Using range()

```python
items = ["a", "b", "c"]

for i in range(len(items)):
    print(items[i])
```

### 2. Using enumerate()

Definition:

enumerate() returns both index and value during iteration.

```python
for index, value in enumerate(items):
    print(index, value)
```

##  Looping Through Collections
Set :

```python
s = {1, 2, 3}

for i in s:
    print(i)
```

Dictionary : 

```python
for key in student:
    print(key)      # printing only the keys in the dictionary....

print()

for value in student.values():
    print(value)    # printing only the value in the dictionary....

print()

for key, value in student.items():
    print(key, value)   # printing both the keys and the corresponding values in the dictionary....
```
