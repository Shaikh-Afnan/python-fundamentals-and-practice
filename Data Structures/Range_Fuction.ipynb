# range() Function in Python
## Definition

range() generates a sequence of numbers starting from a given value up to (but not including) an end value, with an optional step size.

### Why range() is Used

-- To generate numbers for loops

-- To control how many times a loop runs

-- Memory-efficient (does not store all values at once)

## Syntax

### range(start, stop, step)

| Parameter | Meaning                                  |
| --------- | ---------------------------------------- |
| `start`   | Starting number (default = 0)            |
| `stop`    | Ending number (excluded)                 |
| `step`    | Difference between numbers (default = 1) |

### 1. range(stop)
Generates numbers from 0 up to stop - 1.

```python
for i in range(5):
    print(i)
```

### 2. range(start, stop)
Generates numbers starting from start up to stop - 1.

```python
for i in range(2, 7):
    print(i)
```

### 3. range(start, stop, step)
Generates numbers starting from start to stop - 1 with a fixed step value.

```python
for i in range(1, 10, 2):
    print(i)
```

### 4. Using range() with Negative Step
A negative step generates numbers in reverse order.

```python
for i in range(10, 0, -2):
    print(i)
```

### 5. range() with list()
Converts the range object into a list of numbers.

```python
nums = list(range(1, 6))
print(nums)
```

### 6. Important Points
-- range() does not include the stop value

-- range() returns a range object, not a list

-- It is memory efficient

-- Commonly used in for loops
