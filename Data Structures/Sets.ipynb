# Python Sets

A **set** in Python is an unordered collection of unique elements.
Sets automatically remove duplicate values.

### Key Characteristics:
- Unordered
- Mutable
- No duplicate elements
- Elements must be immutable (int, str, tuple)


# 1. Creating sets

Basic Set Concepts

Set Creation – Creating a set means defining an unordered collection of unique elements using {} or set().

Uniqueness – Sets automatically remove duplicate elements.

Empty Set – An empty set is created using set(), not {}.

Iteration – Since sets are unordered, elements are accessed using loops instead of indexing.

```python
# Creating sets
numbers = {1, 2, 3, 4}
names = {"Afnan", "Rahul", "Sara"}
mixed = {1, "Python", 3.14}

print(numbers)
print(names)
print(mixed)
```

### Duplicate Elements Are Removed Automatically
Sets store only unique values; duplicates are removed automatically.

```python
data = {1, 2, 2, 3, 3, 3}
print(data)
```

# 4. Set Methods

### (a) .add(x) 
Adds a single element x to the set.

```python
nums = {1, 2, 3}
nums.add(4)
print(nums)
```

```python
nums.add(5)
nums.add(10)
print(nums)
```

### (b) .update(iterable)
Adds multiple elements from an iterable to the set.

```python
nums.update([6, 7, 8])
print(nums)
```

```python
nums2 = {10,20,30, 90, 100001}
nums2.update([99, 101, 120, 300])
print(nums2)
```

### (c) .remove(x)
Removes element x from the set and raises an error if it is not found.

```python
print(nums)
nums.remove(5)
print(nums)
```

```python
nums.remove(20)    # raises an error because the element does not exists in the set
print(nums)
```

### (d) .discard(x)
Removes element x from the set without raising an error if it is not found.

```python
print(nums)
nums.discard(1)    # removes 1 from the set
print(nums)
```

```python
nums.discard(100)   # does not raises an error
print(nums)
```

### (e) .pop()
Removes and returns a random element from the set.

```python
s1 = {1, 2, 3, 'nit', '1+2j', True, 13.9, 1.62}
removed = s1.pop()
print(removed)
print(s1)
```

```python
removed = s1.pop()
print(removed)
print(s1)
```

```python
removed = s1.pop(2)     # raises an error as the pop() method in set does not accept index as parameter
print(removed)
```

### (f) .clear()
Removes all elements from the set, making it empty.

```python
s2 = {'a', 'b', 'c', 1, 2, 3, 1.2, 3.14, 10, True}
print(s2)
s2.clear()    # removes all the elements from the set
print(s2)     # prints an empty set
```

### (g) .copy() 
Creates a shallow copy of the set.

```python
print(s1)

s3 = s1.copy()

print(s3)
print(s3 == s1)
```

# 5. Set Operations

### (a) .union()
Returns a set containing all unique elements from both sets.

```python
a = {1, 2, 3}
b = {3, 4, 5}

print(a.union(b))
```

```python
c = {'a', 'b', 'c', 'd'}
print(a.union(c))
```

### (b) .intersection()
Returns a set containing elements common to both sets.

```python
print(a)
print(b)
print(a.intersection(b))
print(a.intersection(c))
```

### (c) .difference()
Returns elements that are present in the first set but not in the second set.

```python
print(a)
print(b)
print(a.difference(b))
```

```python
print(a)
print(c)
print(a.difference(c))
```

### (d) .symmetric_difference()
Returns elements that are present in either set but not in both.

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}
c = {5,6,7,11,12,13}
print(a.symmetric_difference(b))
```

```python
print(b.symmetric_difference(c))
```

```python
print(a.symmetric_difference(c))
```

### (e) .update(iterable) 
Adds multiple elements from another iterable (list, tuple, set, etc.) to the set.

```python
new_set = {'a','b', 'c'}
l1 = ['d','e','f']

new_set.update(l1)

print(new_set)   # order can be different.....
```

# 6. In-place Set Operations
In-place set operations modify the original set instead of creating a new one.

### (a) .intersection_update()
Keeps only the elements that are common to both sets and updates the original set.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}

a.intersection_update(b)
print(a)  # Elements 3 and 4 are common, so all other elements are removed from set a.
```

### (b) .difference_update()
Removes elements from the original set that are present in another set.

```python
a = {1, 2, 3, 4}
b = {3, 4}

a.difference_update(b)
print(a)    # Elements 3 and 4 are removed because they exist in set b.
```

### (c) .symmetric_difference_update()
Keeps only elements that are present in either set but not in both, and updates the original set.

```python
a = {1, 2, 3}
b = {3, 4, 5}

a.symmetric_difference_update(b)
print(a)   # Element 3 is removed because it appears in both sets, while unique elements are kept.
```

# 7. Set Relationship Methods in Python
Set relationship methods are used to compare two sets and check how they are related to each other.

### (a) .issubset()
Checks whether all elements of one set are present in another set.

```python
a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))   # Returns True because all elements of set a (1 and 2) exist in set b.
```

```python
print(b.issubset(a))   # Returns False because all elements of set b doesn't exists in set a
```

### (b) .issuperset()
Checks whether a set contains all elements of another set.

```python
a = {1, 2, 3, 4}
b = {2, 3}

print(a.issuperset(b))   # Returns True because set a contains every element of set b.
```

```python
print(b.issuperset(a))   # Returns False because set b does not contain every element of set a 
```

### (c) .isdisjoint()
Checks whether two sets have no elements in common.

```python
x = {1, 2}
y = {3, 4}

print(x.isdisjoint(y))  # Returns True because sets x and y share no common elements.
```

```python
x = {1, 2}
y = {2, 3}

print(x.isdisjoint(y))   # Returns False because sets x and y share atleast 1 element in common.
```

# 8. Built-in Functions with Sets

```python
nums = {10, 20, 30}

print("The length of the set : ",len(nums))     # Returns the number of elements in the set.
print("The largest element in the set : ",max(nums))     # Returns the largest element in the set.
print("The smallest element in the set : ",min(nums))     # Returns the smallest element in the set.
print("The sum of all the elements from set : ",sum(nums))     # Returns the sum of all numeric elements in the set.
```

# Python Sets Mutable or Immutable ?

## Python sets are mutable.

### Short Explanation:

### A set is mutable, which means you can add or remove elements after it is created (using methods like add(), remove(), update()).

### Important Note:

### The set itself is mutable

### But elements inside a set must be immutable (like int, str, tuple)

### One-line interview answer:

### Python sets are mutable collections, but they can contain only immutable elements.
