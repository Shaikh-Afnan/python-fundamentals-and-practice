# Python Bitwise Operators

## PART 1: Number Systems

### Number Systems in Python

A number system defines how numbers are represented using digits.
Python supports multiple number systems internally, but computers work using binary numbers.

### (a) Decimal Number System (Base 10)

Definition :

- The decimal number system uses digits from 0 to 9 and has a base of 10.
- This is the default number system used by humans.

```python
num = 25
print(num)
```

### (b) Binary Number System (Base 2)
Definition :

- The binary number system uses only 0 and 1 and has a base of 2.
- 0b prefix indicates a binary number.

```python
binary_num = 0b1010
print(binary_num)
```

```python
x = 10
print(bin(x))           # bin() function returns the binary value for the given number
```

### (c) Octal Number System (Base 8)
Definition :

- The octal number system uses digits 0 to 7 and has a base of 8.
- 0o prefix indicates an octal number.

```python
x = 10
print(oct(x))       # oct() function returns the octal value for the given number
```

```python
oct_num = 0o12
print(oct_num)
```

### (d) Hexadecimal Number System (Base 16)
Definition

- The hexadecimal number system uses digits 0–9 and A–F and has a base of 16.
- 0x prefix indicates a hexadecimal number.

```python
x = 10
print(hex(x))        # hex() function returns the hexadecimal value for the given number.
```

```python
print(x)
```

```python
x = 100

print(hex(x))
```

```python
print(0x64)
```

### Converting Between Number Systems

```python
num = 10

print("10 in binary : ",bin(num))   # Decimal to Binary
print("10 in octal : ",oct(num))   # Decimal to Octal
print("10 in hexadecimal : ",hex(num))   # Decimal to Hexadecimal
```

## PART 2: Bitwise Operators in Python

### What Are Bitwise Operators?
Bitwise operators perform operations directly on the binary representation of numbers.
Each bit (0 or 1) is processed individually.
### Types of Bitwise Operators :

### (a) Bitwise AND (&)
Definition :

- Returns 1 if both corresponding bits are 1; otherwise returns 0.

```python
a = 5    # 101
b = 3    # 011

print(a & b)
```

  101
& 011
-----
 = 001  = 1

### (b) Bitwise OR (|)
Definition :

- Returns 1 if at least one corresponding bit is 1.

```python
a = 5    # 101
b = 3    # 011

print(a | b)
```

  101
| 011
-----
 = 111  = 7

### (c) Bitwise XOR (^)
Definition :

- Returns 1 if corresponding bits are different; otherwise returns 0.

```python
a = 5
b = 3

print(a ^ b)
```

  101
^ 011
-----
 = 110  = 6

### (d) Bitwise NOT (~) or Complement Operator
Definition :

- Inverts all the bits (0 becomes 1 and 1 becomes 0).

```python
x = 5
print(~x)
```

- ~5 = -(5 + 1) = -6
- Python uses 2’s complement for negative numbers.

### (e) Left Shift (<<)
Definition :

- Shifts bits to the left, filling with zeros on the right.

```python
x = 5
print(x << 1)
```

5 → 101

Left shift by 1 → 1010 → 10

Equivalent to multiplying by 2.

### (f) Right Shift (>>)
Definition :

- Shifts bits to the right, removing bits from the right side.

```python
x = 10
print(x >> 1)
```

10 → 1010

Right shift by 1 → 0101 → 5

Equivalent to dividing by 2.
