# Python Dictionary

A **dictionary** in Python is an unordered, mutable collection of data
that stores values in **key : value** pairs.

### Key Characteristics:
- Stores data as key-value pairs
- Keys must be unique and immutable
- Values can be of any data type
- Dictionaries are mutable


# 1. Creating Dictionaries
Dictionary creation means defining key–value pairs using curly braces {} or the dict() constructor.

```python
# Creating an empty dictionary......

mydict = {}
mydict2 = dict()

print(type(mydict))
print(type(mydict2))
```

```python
# Creating a dictionary with key value pairs.....
student = {
    "name": "John",
    "age": 22,
    "course": "Computer Science",
    "marks": 85
}

print(type(student))
print(student)
```

# 2. Accessing Dictionary Values
There are two ways to access dictionary values
1. By using square brackets ([]).
2. By using .get() method

```python
print(student["name"])
print(student.get("course"))
```

### Important Note:
- [] raises error if key is missing
- get() returns None instead of error

```python
print(student["lastname"])    # raises an error
```

```python
print(student.get("lastname"))    # Returns None
```

# 3. Modifying Dictionary Values
Dictionary modification allows updating existing values or adding new key–value pairs.

```python
student["age"] = 23           # updating the value of 'age' key from 22 to 23
student["city"] = "Mumbai"    # adding a new key value pair to the student dictionary

print(student)
```

# 4. Removing Elements from Dictionary
Removing elements deletes a key–value pair from the dictionary using methods like pop().

```python
student.pop("marks")     # removes the key value pair "marks" : 85 from the student dictionary
print(student)
```

# 5. Dictionary Methods

### (a) .get() 
Returns the value of a given key, or None if the key does not exist.

```python
print(student.get("name"))
print(student.get("grade"))   # returns None
```

### (b) .keys()
Returns a view object containing all the keys in the dictionary.

```python
print(student.keys())
```

### (c) .values()
Returns a view object containing all the values in the dictionary.

```python
print(student.values())
```

### (d) .items()
Returns a view object containing key–value pairs as tuples.

```python
print(student.items())
```

### (e) .update()
Adds or updates key–value pairs from another dictionary or iterable.

```python
student.update({"grade": "A", "year": 2024})
print(student)
```

### (f) .pop()
Removes the specified key and returns its associated value or raises an error if the key doesn't exists in the dictionary.

```python
removed = student.pop("city")
print(removed)
print(student)
```

```python
removed = student.pop("lastname")    # raises an error because "lastname" is not
print(removed)                       # present in the dictionary
print(student)
```

### (g) .popitem()
Removes and returns the last inserted key–value pair from the dictionary.

```python
removed_item = student.popitem()
print(removed_item)
print(student)
```

### (h) .copy()
Creates a shallow copy of the dictionary.

```python
student_copy = student.copy()
print("Student : ",student)
print("Copy of the Student : ",student_copy)
```

### (i) .clear()
Removes all key–value pairs, making the dictionary empty.

```python
temp = {"a": 1, "b": 2}
print(temp)

temp.clear()
print(temp)
```

### (j) .fromkeys()
Creates a new dictionary using given keys with a single common value.

```python
keys = ["name", "age", "course"]
new_dict = dict.fromkeys(keys, None)
print(new_dict)
```

# 6. Looping Through Dictionary
Iteration allows accessing keys, values, or key–value pairs using loops.

### (a) Loop through keys

```python
for key in student:
    print(key)          # prints all the keys from the dictionary.....
```

### (b) Loop through values

```python
for value in student.values():
    print(value)
```

### (c) Loop through key-value pair

```python
for key, value in student.items():
    print(key," : ", value)
```

# 7. Dictionary Comprehension
A concise way to create dictionaries using expressions and loops.

```python
squares = {x: x**2 for x in range(1, 6)}

print(squares)
```

# 8. Nested Dictionary
A dictionary that contains another dictionary as its value.

```python
students = {
    "student1": {"name": "John", "marks": 85},
    "student2": {"name": "Irwin", "marks": 90}
}

print(students["student1"])
print(students["student2"])

print(students["student1"]["name"])
print(students["student2"]["name"])
```

# 9. Dictionary Keys Rules
### Key Rules:
- Keys must be unique
- Keys must be immutable (int, str, tuple)
- Lists and sets cannot be keys

# 10. Common Mistakes when using Dictionaries

- Using mutable objects as keys
- Forgetting keys must be unique
- Confusing `get()` with indexing
