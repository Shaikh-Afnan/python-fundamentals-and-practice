# 1. Introduction to Tuples

# Python Tuples

A **tuple** in Python is an ordered, immutable collection of elements.
Once created, a tuple cannot be modified.

### Key Characteristics:
- Ordered
- Immutable (cannot be changed)
- Allows duplicate elements
- Indexed (starts from 0)

# 2. Creating Tuples 
Creating a tuple means defining an ordered, immutable collection using parentheses ().

```python
# Creating tuples
numbers = (10, 20, 30)      # 
names = ("Alice", "Bob", "Charlie")
mixed = (1, "Python", 3.14, True)

print(numbers)
print(names)
print(mixed)
```

# 3. Single-Element Tuple
A tuple with only one element must include a trailing comma, otherwise it is not treated as a tuple.

```python
# Correct way to create a single-element tuple
single = (10,)
print(single)
print(type(single))
```

```python
# This is NOT a tuple
not_a_tuple = (10)
print(type(not_a_tuple))
```

# 4. Accessing Tuple Elements
Tuple elements are accessed using index positions starting from 0.

```python
fruits = ("apple", "banana", "mango", "orange")

print(fruits[0])
print(fruits[-1])       # Negative indexes allow access to tuple elements starting from the end.
```

# 5. Tuple Slicing
Slicing extracts a portion of a tuple using a start, stop, and step.


```python
numbers = (0, 1, 2, 3, 4, 5, 6)

print("The numbers from index 1 to 4 are : ",numbers[1:5])
print("The numbers from starting index to 4 are : ",numbers[:4])
print("The numbers from index 3 to last are : ",numbers[3:])
print("The numbers from index 0 to last in the steps of 2 are : ",numbers[::2])
```

# 6. Immutability of Tuples
Tuples cannot be modified, meaning elements cannot be added, removed, or changed after creation.

```python
data = (10, 20, 30)

# This will raise an error
data[0] = 100
```

# 7. Tuple Methods

## (a) .count(x)
Returns the number of times element x appears in the tuple.

```python
nums = (1, 2, 2, 3, 2)
print(nums.count(2))
```

## (b) .index(x)
It is similar to the list index() method. It Returns the index position of the first occurrence of element x in the tuple.

```python
names = ("Afnan", "Rahul", "Sara", "Atharva", "Jenny", "Sara")
print(names.index("Sara"))
```

# 8. Built-in Functions with Tuples
Functions like len(), max(), min(), and sum() operate on tuples to retrieve information.

```python
nums = (10, 20, 30, 40)

print(len(nums))
print(max(nums))
print(min(nums))
print(sum(nums))
```

# 9. Tuple Packing and Unpacking
Tuple Packing – Packing means grouping multiple values into a single tuple.

Tuple Unpacking – Unpacking assigns tuple elements to multiple variables in one statement.

```python
# Packing
student = ("Afnan", 22, "MCA")

# Unpacking
name, age, course = student

print(name)
print(age)
print(course)
```

# 10. Tuple Unpacking in Loop
Tuple unpacking inside loops allows accessing multiple values directly from tuple elements.

```python
students = [("Afnan", 90), ("Rahul", 85), ("Sara", 92)]

for name, marks in students:
    print(name, "scored", marks)
```

# 11. Tuple to List Conversion(To Modify Data)
Converting a tuple to a list allows modification since lists are mutable.

```python
data = (10, 20, 30)

temp = list(data)
temp.append(40)

data = tuple(temp)
print(data)
```

# Conclusion
### Key Points:
- Tuples are immutable
- Faster than lists
- Used for fixed data
- Supports only `count()` and `index()`
