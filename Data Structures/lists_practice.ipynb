# Intro to Python Lists

A **list** in Python is an ordered, mutable collection of items.
Lists can store multiple data types such as integers, strings, floats, and even other lists.

### Key Characteristics:
- Ordered
- Mutable (can be changed)
- Allows duplicate elements
- Indexed (starts from 0)


### 1. Creating Lists :
Lists are represented using the square brackets i.e. the elements are enclosed in "[]" Brackets

```python
l = []     # creating an empty list
l2 = [1,2,3,4,5,6]  # defining a list with elements
print(type(l))
print(type(l2))
```

```python
# We can add elements in the list using append method which add elements to the end of the list

l.append(10)
l.append(20)
l.append(30)
l
```

```python
# We can add elements of different data types in a list
l.append('any string')
l.append(2.13)
l.append(True)
l.append(1+2j)
l.append(None)
l
```

### 2. Accessing the List elements 
List elements can be accessed using index positions which always start from 0

```python
l[0] # accessing the 1st element
```

```python
print(l[3]) # accessing the 4th element
print(l[4]) # acessing the 5th element
```

### 3. List Slicing

List slicing in Python is a technique used to extract a specific portion, or sublist, of elements from an existing list. It provides a concise way to access a range of elements by specifying a start, stop, and optional step size using a colon (:) notation within square brackets. 


```python
print("No. 1 : " ,l[0:3]) # printing the sub list starting from 0 index to index 2
print("No. 2 : " ,l[:]) # printing all of the list
print("No. 3 : " ,l[:5]) # printing the sub list starting from 0 index to index 4
print("No. 4 : " ,l[3:])  # printing the sub list starting from 3 index to the last index
print("No. 5 : " ,l[::2]) # printing the sub list from start to end in the steps of 2
print("No. 6 : " ,l[::-1]) # printing the list in reverse 
print("No. 7 : " ,l[::-2]) # printing the list in reverse in steps of 2
```

### 4. Modifying List elements
Elements in the list can be modified using the square brackets to access them and assigning new value to them.

```python
print("Original Value : ", l[2])
print("list : ", l)
l[2] = 200
print("Changed Value : ", l[2])
print("list with Changed Value : ",l)
```

### 5. Important List Methods :

#### (a) .append(x)
Adds a single element x to the end of the list.

```python
print(l2)
l2.append(10) # only 1 element can be appended at a time.
l2.append(True)
print(l2)
```

```python
l2.append(12,13) # results in an error
```

#### (b) .clear()
Removes all elements from the list, making it empty.

```python
print(l2)
l2.clear()
print(l2)
```

#### (c) .remove(x)
Removes the first occurrence of element x from the list.

```python
l2 = [10,20,30,40,50,10,20,15]
l2.remove(10)
print(l2)
```

```python
l2.remove(15)
print(l2)
```

#### (d) .pop([index])
Removes and returns the element at the given index (last element by default).

```python
l2.pop() # removes the last element by default
```

```python
l2.pop(2)
```

```python
print(l2)
```

#### (e) .insert(index, x)
Inserts an element x at the specified index position.

```python
l2.insert(2, 45)
l2.insert(1, 100) 
print(l2)
```

#### (f) .extend(iterable)
Adds all elements from another iterable to the end of the list.

```python
l3 = ['a', 'b', 'c', 'd']
l2.extend(l3)
print(l2)
```

#### (g) .reverse()
Reverses the order of elements in the list in place.

```python
print("Before reverse : ", l2)
l2.reverse() # the entire list object is reversed 
print("After reverse : ",l2)
```

#### (h) .sort()
Sorts the list elements in ascending order by default.

```python
l4 = [20, 10, 30, 5, 2, -1, 100, 50]   # contains only numbers
print("Before Sorting : ", l4)
l4.sort()
print("After Sort : ", l4)
```

```python
# sort() does not apply on lists with heterogeneous type of elements
print(l2)   # Contains elements of different data types
l2.sort()   # Throws an error..... 
```

#### we can also sort a list in descending order by using reverse = True parameter in sort() method......

```python
l5 = ['b', 'd', 'g', 'a', 'c', 'q', 'f']    
l5.sort(reverse= True)
print(l5)
```

#### (i) .count(x)
Returns the number of times element x appears in the list.

```python
l6 = ['a', 10, 22, 40, 'a', 30, 'a', 22]
print("Count of 'a' in l6 : ", l6.count('a'))
print("Count of 22 in l6 : ", l6.count(22))
```

#### (j) .index(x)
Returns the index of the first occurrence of element x.

```python
print("index of 'a' in l6 : ", l6.index('a'))   # returns 0 as the first occurence of 'a' is at index 0
print("Index of 22 in l6 : ", l6.index(22))     # returns 2 as the first occurence of 22 is at index 2
```

### 6. Built-in Functions of Lists

```python
l7 = [99, 100, 10, 9, 20, 40, 1110]
print('The length of the list l7 is : ', len(l7))    # returns the total count of the elements i.e. length of the list
print('The Biggest number in the list l7 is : ', max(l7))  #returns the biggest element from the list
print('The Smallest number in the list l7 is : ', min(l7))  #returns the smallest element from the list
print('The Sum Total of all elements in the list l7 is : ', sum(l7))  #returns the sum total of all element from the list
```

# Reference Copy vs Absolute Copy in Python Lists

In Python, assigning a list to another variable does NOT create a new list.
Instead, it creates a reference to the same list object in memory.

The Code Below explains:
- Reference Copy
- Absolute Copy (Shallow Copy)
- How memory behaves in both cases


## 1. Reference Copy (Assignment) – SAME MEMORY
A reference copy means both variables point to the **same list object in memory**.
Any modification through one variable affects the other.

```python
# Reference copy example
list_a = [10, 20, 30]
list_b = list_a   # reference copy

print("List A:", list_a)
print("List B:", list_b)

print("Memory Address of List A:", id(list_a))
print("Memory Address of List B:", id(list_b))
```

```python
# Modify one list :
list_b.append(40)

print("After modification:")
print("List A:", list_a)
print("List B:", list_b)
```

### Observation:
- Both lists have the same memory address
- Changing one list changes the other
- This is NOT a true copy


## 2. Absolute Copy (Shallow Copy) – DIFFERENT MEMORY
An absolute copy creates a **new list object** with the same elements.
Changes in one list do NOT affect the other.

### (a) Method 1: Using copy() Method

```python
list_x = [1, 2, 3]
list_y = list_x.copy()

print("List X:", list_x)
print("List Y:", list_y)

print("Memory Address of List X:", id(list_x))
print("Memory Address of List Y:", id(list_y))
```

```python
# Modify Copied List
list_y.append(4)

print("After modification:")
print("List X:", list_x)
print("List Y:", list_y)
```

### (b) Method 2: Using Slicing [:]

```python
list_p = [100, 200, 300]
list_q = list_p[:]

print("List P:", list_p)
print("List Q:", list_q)

print("Memory Address of List P:", id(list_p))
print("Memory Address of List Q:", id(list_q))
```

### (c) Method 3: Using list() Constructor

```python
list_m = [7, 8, 9]
list_n = list(list_m)

print("List M:", list_m)
print("List N:", list_n)

print("Memory Address of List M:", id(list_m))
print("Memory Address of List N:", id(list_n))
```
