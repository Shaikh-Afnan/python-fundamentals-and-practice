#  Data Cleaning or Data Cleansing.....

## Step 5: Cleaning Text Data

Special characters were removed from text columns such as:

- Name
- Domain
- Salary
- Experience
- Age

We used `str.replace()` with regular expressions to remove:

- Non-word characters (`\W`)
- Non-digit characters (`\D`)

This ensures consistent and clean textual data.


```python
emp
```

```python
emp['Name']
```

```python
emp['Name'] = emp['Name'].str.replace(r'\W', '', regex = True)   # Non Word Character
```

```python
emp['Name']
```

```python
emp.dtypes
```

```python
emp['Domain'] = emp['Domain'].str.replace(r'\W', '', regex = True)
emp['Domain']
```

```python
emp.head()
```

## Step 6: Data Type Conversion

Some columns were stored as strings but should be numeric.

We converted:
- Salary → int
- Experience → float → int
- Age → float → int

Using `astype()` ensures correct data types for analysis and calculations.

```python
emp['Salary'] = emp['Salary'].str.replace(r'\W', '', regex = True)

emp['Salary']
```

```python
emp['Salary'] = emp['Salary'].astype(int)
emp['Salary'].dtypes
```

```python
emp['Exp'] = emp['Exp'].str.replace(r'\D', '', regex= True)
emp['Exp']
```

```python
emp['Exp'].isnull().sum()
```

```python
emp['Exp'] = emp['Exp'].astype('float')
```

## Step 7: Filling Missing Numerical Values

For numerical columns such as:

- Experience
- Age

Missing values were replaced with the column mean using:

`replace(np.nan, column.mean())`

This prevents data loss and maintains statistical balance.


```python
import numpy as np
emp['Exp'] = emp['Exp'].replace(np.nan, emp['Exp'].mean())
emp
```

```python
emp['Exp'] = emp['Exp'].astype('int')

emp
```

```python
emp['Age'] = emp['Age'].str.replace(r'\D', '', regex = True)

emp['Age']
```

```python
emp['Age'] = emp['Age'].astype('float')
emp['Age'].dtype
```

```python
emp["Age"] = emp['Age'].replace(np.nan, emp['Age'].mean())
emp
```

```python
emp['Age'] = emp['Age'].astype('int')
emp
```

## Step 8: Standardizing Categorical Data

Some categorical values were inconsistent, such as:

- "Datascience" → "Data Science"
- "Dataanalyst" → "Data Analyst"

We replaced them using `.replace()` to maintain uniform category naming.

This improves grouping and analysis accuracy.


```python
emp['Domain'] = emp['Domain'].replace('Datascience', 'Data Science').replace('Dataanalyst', 'Data Analyst')
emp
```

## Step 9: Filling Missing Categorical Values

For the 'Location' column:

Missing values were replaced with "Unknown".

This ensures:
- No null values remain
- Categorical consistency
- Better grouping during analysis

```python
print(emp['Location'].isnull().sum())
```

```python
emp['Location'] = emp['Location'].replace(np.nan, 'Unknown')

emp
```
